```python
import sys
from pathlib import Path
import pandas as pd

REPO_ROOT = Path('..').resolve()
sys.path.insert(0, str(REPO_ROOT / 'src'))
from jnpdi_validator import validate_dataset

data_path = REPO_ROOT / 'datasets/03_business_economic_justice/data.csv'
sources_path = REPO_ROOT / 'datasets/03_business_economic_justice/sources.csv'
indicators_path = REPO_ROOT / 'metadata/indicators_50.csv'

report = validate_dataset(data_path, indicators_path, sources_path)
print('Validation Passed:', report.passed, '| Rows:', report.row_count)
assert report.passed, report.errors
```

```python
df = pd.read_csv(data_path)
assert (df['classification'] == 'P5').all()
assert (df['is_synthetic'].astype(str).str.lower() == 'true').all()
summary = df.groupby(['indicator_id', 'year'])['value'].mean().unstack()
display(summary.head(10))
```
